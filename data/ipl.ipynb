```python
balls_faced=deliveries.groupby("striker")["over"].count()
```

```python
balls_faced = balls_faced.rename(columns={'over': 'balls_faced'})
```

```python
batter_stats=batters_runs.merge(balls_faced,on="striker")
```

```python
batter_stats['strike_rate'] = (
    batter_stats['runs_of_bat'] /
    batter_stats['balls_faced']
) * 100
```

```python
# Bowling Analysis
#Top wickets takers 
wickets=deliveries[
    deliveries["player_dismissed"].notna()
    ]
    
```

```python
wickets.head(10)
```

```python
top_wickets=wickets.groupby("bowler")["player_dismissed"].count().reset_index()
```

```python
top_wickets=top_wickets.rename(
    columns={"player_dismissed":"wickets"}
)
```

```python
top_wickets.head(10)
```

```python
top_wickets=top_wickets.sort_values(
    by="wickets",
    ascending=False
)
```

```python
top_wickets.head(10)
```

```python
deliveries['bowler_runs'] = (
    deliveries['runs_of_bat'] +
    deliveries['wide'] +
    deliveries['noballs']
)
```

```python
#runs conceeded
runs_conceded=deliveries.groupby("bowler")["bowler_runs"].sum().reset_index()
```

```python
runs_conceded.head(10)
```

```python
legal_deliveries=deliveries[
    (deliveries["wide"]==0)&
    (deliveries["noballs"]==0)
    ]
```

```python
balls_bowled=legal_deliveries.groupby("bowler")["over"].count().reset_index()
```

```python
balls_bowled=balls_bowled.rename(
    columns={"over": "balls_bowled"}
)
```

```python
balls_bowled["overs_bowled"]=(
    balls_bowled["balls_bowled"]/6
)
```

```python
balls_bowled.head(10)
```

```python
economy=runs_conceded.merge(
    balls_bowled,
    on="bowler"
)
```

```python
economy["economy_rate"]=(
    economy["bowler_runs"]/
    economy["overs_bowled"]
)
```

```python
economy.head(10)
```

```python
economy_filtered = economy[
    economy['overs_bowled'] >= 2
]
```

```python
economy_filtered.head(10)
```

```python
economy_filtered=economy_filtered.sort_values(
    by="economy_rate",
    ascending=True
)
```

```python
economy_filtered.head(10)
```

```python
#Dot ball analysis
dot_balls = deliveries[
    (deliveries['runs_of_bat'] == 0) &
    (deliveries['wide'] == 0) &
    (deliveries['noballs'] == 0)
]
```

```python
dot_balls_stats=dot_balls.groupby("bowler")["over"].count().reset_index()
```

```python
dot_balls_stats=dot_balls_stats.rename(
    columns={"over":"dot_balls"}
)
```

```python
dot_balls_stats.head(10)
```

```python
dot_balls_stats=dot_balls_stats.sort_values(
    by="dot_balls",
    ascending=False
)
```

```python
dot_balls_stats.head(10)
```

```python
#Death balls analysis
death_overs=deliveries[
    deliveries["over"]>=16
    ].copy()
```

```python
death_overs['death_runs'] = (
    death_overs['runs_of_bat'] +
    death_overs['wide'] +
    death_overs['noballs']
)
```

```python
deliveries.info()
```

```python
death_runs = death_overs.groupby('bowler')[
    'death_runs'
].sum().reset_index()
```

```python
death_runs.head(10)
```

```python
death_legal = death_overs[
    (death_overs['wide'] == 0) &
    (death_overs['noballs'] == 0)
]
```

```python
death_balls=death_legal.groupby("bowler")["over"].count().reset_index()
```

```python
death_balls=death_balls.rename(
    columns={"over":"death_balls"}
)
```

```python
death_balls["death_overs"]=(
    death_balls["death_balls"]/6
)
```

```python
death_balls.head(10)
```

```python
death_stats=death_runs.merge(
    death_balls,
    on="bowler"
)
```

```python
death_stats.head(10)
```

```python
death_stats["death_economy"]=(
    death_stats["death_runs"]/
    death_stats["death_overs"]
)
```

```python
death_stats.head(10)
```

```python
death_stats=death_stats.sort_values(
    by="death_economy",
    ascending=True
)
```

```python
death_stats.head(10)
```

```python
death_stats = death_stats[
    death_stats['death_overs'] >= 2
]
```

```python
death_stats = death_stats.sort_values(
    by='death_economy'
)
```

```python
death_stats.head(10)
```

```python
#chasing friendly grounds 
matches["chasing_wins"]=(
    matches["match_winner"]==matches["team2"]
)
```

```python
matches[['team1','team2','match_winner','chasing_wins']].head()
```

```python
matches['chasing_team'] = np.where(

    matches['toss_decision'] == 'bat',

    np.where(
        matches['toss_winner'] == matches['team1'],
        matches['team2'],
        matches['team1']
    ),

    matches['toss_winner']
)
```

```python
matches['chasing_wins'] = (
    matches['match_winner'] ==
    matches['chasing_team']
)
```

```python
#Toss impact analysis
matches['toss_win_match_win'] = (
    matches['toss_winner'] ==
    matches['match_winner']
)
```

```python
toss_impact = (
    matches['toss_win_match_win']
    .mean()
) * 100
```

```python
print(
    "Toss winners also won match:",
    round(toss_impact,2),
    "%"
)
```

```python
matches[['toss_winner','match_winner','toss_win_match_win']].head(10)
```

```python
#Toss winners won 53.85% of matches, suggesting toss advantage exists but is not a dominant match-winning factor in this dataset.
```

```python
chasing_wins = matches['chasing_win'].sum()
```

```python
total_matches = matches.shape[0]
```

```python
batting_first_wins = (
    total_matches - chasing_wins
)
```

```python
chasing_win_percent = (
    chasing_wins / total_matches
) * 100
```

```python
batting_first_win_percent = (
    batting_first_wins / total_matches
) * 100
```

```python
print(
    "Chasing win percentage:",
    round(chasing_win_percent,2),
    "%"
)

print(
    "Batting first win percentage:",
    round(batting_first_win_percent,2),
    "%"
)
```
