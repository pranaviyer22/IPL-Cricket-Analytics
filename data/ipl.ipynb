```python
balls_faced=deliveries.groupby("striker")["over"].count()
```

```python
balls_faced=balls_faced.reset_index()
```

```python
balls_faced.head()
```

```python
balls_faced = balls_faced.rename(columns={'over': 'balls_faced'})
```

```python
balls_faced.head(10)
```

```python
batters_runs = deliveries.groupby('striker')['runs_of_bat'].sum().reset_index()
```

```python
batter_stats=batters_runs.merge(balls_faced,on="striker")
```

```python
batter_stats['strike_rate'] = (
    batter_stats['runs_of_bat'] /
    batter_stats['balls_faced']
) * 100
```

```python
batters_runs.head(10)
```

```python
batter_stats.head()
```

```python
top_strike_rates=batter_stats[
    batter_stats["balls_faced"]>=50
    ]
```

```python
top_strike_rates=top_strike_rates.sort_values(
    by="strike_rate",
    ascending=False
)
```

```python
top_strike_rates.head()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x="striker",
    y="strike_rate",
    data=top_strike_rates.head(10)
)

plt.title("Top strike rate batters")
plt.xlabel("Batters")
plt.ylabel("Strike Rate")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fours=deliveries[
    deliveries["runs_of_bat"]==4
]
```

```python
top_fours=fours.groupby("striker")["runs_of_bat"].count()
```

```python
top_fours.head(10)
```

```python
top_fours=top_fours.reset_index()
```

```python
top_fours=top_fours.rename(
    columns={"runs_of_bat":"fours"}
) 
    
```

```python
top_fours=top_fours.sort_values(
    by="fours",
    ascending=False
)
```

```python
top_fours.head(10)
```

```python
sixes=deliveries[
    deliveries["runs_of_bat"]==6
]
```

```python
top_sixes=sixes.groupby("striker")["runs_of_bat"].count().reset_index()
```

```python
top_sixes.head(10)
```

```python
top_sixes=top_sixes.rename(
    columns={"runs_of_bat":"sixes"}
)
```

```python
top_sixes=top_sixes.sort_values(
    by="sixes",
    ascending=False
)
```

```python
top_sixes.head(10)
```

```python
batter_stats = batter_stats.merge(
    top_fours,
    on='striker',
    how='left'
)
```

```python
batter_stats.head(10)
```

```python
batter_stats = batter_stats.merge(
    top_sixes,
    on='striker',
    how='left'
)
```

```python
batter_stats.head(10)
```

```python
batter_stats[['fours','sixes']]=batter_stats[
    ['fours','sixes']
    ].fillna(0)
```

```python
batter_stats['boundary_runs'] = (
    batter_stats['fours'] * 4
) + (
    batter_stats['sixes'] * 6
)
```

```python
batter_stats['boundary_dependency'] = (
    batter_stats['boundary_runs'] /
    batter_stats['runs_of_bat']
) * 100
```

```python
batter_stats.head(10)
```

```python
batter_stats['fours'] = batter_stats['fours'].astype(int)

batter_stats['sixes'] = batter_stats['sixes'].astype(int)
batter_stats["boundary_runs"]=batter_stats["boundary_runs"].astype(int)
```

```python
batter_stats.head(10)
```

```python
batter_stats['boundary_dependency'] = batter_stats['boundary_dependency'].fillna(0)
```

```python
batter_stats.head(10)
```

```python
top_boundary_dependency=batter_stats[
    batter_stats["balls_faced"]>=50
    ]
```

```python
top_boundary_dependency=top_boundary_dependency.sort_values(
    by="boundary_dependency",
    ascending=False
)
```

```python
top_boundary_dependency.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='striker',
    y='boundary_dependency',
    data=top_boundary_dependency.head(10)
)

plt.title("Top Boundary Dependent Batters")
plt.xlabel("Batters")
plt.ylabel("Boundary Dependency %")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
batter_stats.to_csv(
    "final_batter_analysis.csv",
    index=False
)
```

```python
# Bowling Analysis
#Top wickets takers 
wickets=deliveries[
    deliveries["player_dismissed"].notna()
    ]
    
```

```python
wickets.head(10)
```

```python
top_wickets=wickets.groupby("bowler")["player_dismissed"].count().reset_index()
```

```python
top_wickets=top_wickets.rename(
    columns={"player_dismissed":"wickets"}
)
```

```python
top_wickets.head(10)
```

```python
top_wickets=top_wickets.sort_values(
    by="wickets",
    ascending=False
)
```

```python
top_wickets.head(10)
```

```python
deliveries['bowler_runs'] = (
    deliveries['runs_of_bat'] +
    deliveries['wide'] +
    deliveries['noballs']
)
```

```python
#runs conceeded
runs_conceded=deliveries.groupby("bowler")["bowler_runs"].sum().reset_index()
```

```python
runs_conceded.head(10)
```

```python
legal_deliveries=deliveries[
    (deliveries["wide"]==0)&
    (deliveries["noballs"]==0)
    ]
```

```python
balls_bowled=legal_deliveries.groupby("bowler")["over"].count().reset_index()
```

```python
balls_bowled=balls_bowled.rename(
    columns={"over": "balls_bowled"}
)
```

```python
balls_bowled["overs_bowled"]=(
    balls_bowled["balls_bowled"]/6
)
```

```python
balls_bowled.head(10)
```

```python
economy=runs_conceded.merge(
    balls_bowled,
    on="bowler"
)
```

```python
economy["economy_rate"]=(
    economy["bowler_runs"]/
    economy["overs_bowled"]
)
```

```python
economy.head(10)
```

```python
economy_filtered = economy[
    economy['overs_bowled'] >= 2
]
```

```python
economy_filtered.head(10)
```

```python
economy_filtered=economy_filtered.sort_values(
    by="economy_rate",
    ascending=True
)
```

```python
economy_filtered.head(10)
```

```python
#Dot ball analysis
dot_balls = deliveries[
    (deliveries['runs_of_bat'] == 0) &
    (deliveries['wide'] == 0) &
    (deliveries['noballs'] == 0)
]
```

```python
dot_balls_stats=dot_balls.groupby("bowler")["over"].count().reset_index()
```

```python
dot_balls_stats=dot_balls_stats.rename(
    columns={"over":"dot_balls"}
)
```

```python
dot_balls_stats.head(10)
```

```python
dot_balls_stats=dot_balls_stats.sort_values(
    by="dot_balls",
    ascending=False
)
```

```python
dot_balls_stats.head(10)
```

```python
#Death balls analysis
death_overs=deliveries[
    deliveries["over"]>=16
    ].copy()
```

```python
death_overs['death_runs'] = (
    death_overs['runs_of_bat'] +
    death_overs['wide'] +
    death_overs['noballs']
)
```

```python
deliveries.info()
```

```python
death_runs = death_overs.groupby('bowler')[
    'death_runs'
].sum().reset_index()
```

```python
death_runs.head(10)
```

```python
death_legal = death_overs[
    (death_overs['wide'] == 0) &
    (death_overs['noballs'] == 0)
]
```

```python
death_balls=death_legal.groupby("bowler")["over"].count().reset_index()
```

```python
death_balls=death_balls.rename(
    columns={"over":"death_balls"}
)
```

```python
death_balls["death_overs"]=(
    death_balls["death_balls"]/6
)
```

```python
death_balls.head(10)
```

```python
death_stats=death_runs.merge(
    death_balls,
    on="bowler"
)
```

```python
death_stats.head(10)
```

```python
death_stats["death_economy"]=(
    death_stats["death_runs"]/
    death_stats["death_overs"]
)
```

```python
death_stats.head(10)
```

```python
death_stats=death_stats.sort_values(
    by="death_economy",
    ascending=True
)
```

```python
death_stats.head(10)
```

```python
death_stats = death_stats[
    death_stats['death_overs'] >= 2
]
```

```python
death_stats = death_stats.sort_values(
    by='death_economy'
)
```

```python
death_stats.head(10)
```

```python
bowler_stats = top_wickets.merge(
    economy_filtered[
        ['bowler','economy_rate']
    ],
    on='bowler'
)
```

```python
bowler_stats = bowler_stats.merge(
    dot_balls_stats,
    on='bowler'
)
```

```python
bowler_stats = bowler_stats.merge(
    death_stats[
        ['bowler','death_economy']
    ],
    on='bowler'
)
```

```python
bowler_stats.head(10)
```

```python
bowler_stats = bowler_stats.sort_values(
    by='wickets',
    ascending=False
)
```

```python
bowler_stats.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='bowler',
    y='wickets',
    data=bowler_stats.head(10)
)

plt.title("Top Wicket Takers")
plt.xlabel("Bowlers")
plt.ylabel("Wickets")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='bowler',
    y='economy_rate',
    data=economy_filtered.head(10)
)

plt.title("Best Economy Bowlers")
plt.xlabel("Bowlers")
plt.ylabel("Economy Rate")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='bowler',
    y='dot_balls',
    data=dot_balls_stats.head(10)
)

plt.title("Top Dot Ball Bowlers")
plt.xlabel("Bowlers")
plt.ylabel("Dot Balls")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='bowler',
    y='death_economy',
    data=death_stats.head(10)
)

plt.title("Best Death Over Bowlers")
plt.xlabel("Bowlers")
plt.ylabel("Death Economy")

plt.xticks(rotation=45)

plt.show()
```
