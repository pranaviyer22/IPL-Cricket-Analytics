```python
# WHICH TEAM WON MOST MATCHES?
matches["match_winner"].value_counts()
```

```python
import matplotlib.pyplot as plt
```

```python
wins=matches["match_winner"].value_counts()
```

```python
# “Which team dominated this season?”
wins.plot(kind="bar")
plt.title("Matches won by each team")
plt.xlabel("Teams")
plt.ylabel("Number of wins")
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    x=wins.index,
    y=wins.values
)
plt.title("Matches won by each team",fontsize=16)
plt.xlabel("Team",fontsize=14)
plt.ylabel("Wins",fontsize=14)


plt.show()
```

```python
# How many runs did each batter score?
batters_runs=deliveries.groupby("striker")['runs_of_bat'].sum()
```

```python
batters_runs.sort_values(ascending=False).head(10)
```

```python
batters_runs=batters_runs.reset_index()
```

```python
batters_runs.head()
```

```python
Top_batters=batters_runs.sort_values(by='runs_of_bat',ascending=False).head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x='striker',
    y='runs_of_bat',
    data=Top_batters
)
plt.title("Top 10 run getter")
plt.xlabel("Batters")
plt.ylabel("Runs")

plt.xticks(rotation=45)
plt.show()
   
```

```python
Top_batters=Top_batters.reset_index()
```
